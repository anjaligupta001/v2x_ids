```python
from pathlib import Path
import json, re
import pandas as pd
```

```python
DATA = Path.home() / "Documents/v2x-ids/data/raw/mixall_sample/run_79200"
files = sorted(DATA.glob("traceJSON-*.json"))
print("Number of car files:", len(files))
```

```python
print(files[0].name)
m = re.match(r"traceJSON-(\d+)-(\d+)-A(\d+)-(\d+)-(\d+)\.json", files[0].name)
print("vehicle id:", m[1], "| attack code:", m[3], "| start time:", m[4])
```

```python
f = next(f for f in files if "-A0-" in f.name)
df = pd.DataFrame([json.loads(line) for line in open(f)])
df["type"].value_counts()
```

```python
recv = df[df["type"] == 3].copy()
recv["sender"] = recv["sender"].astype(int)
recv["x"] = recv["pos"].str[0]
recv["y"] = recv["pos"].str[1]
recv["speed"] = recv["spd"].apply(lambda v: (v[0]**2 + v[1]**2) ** 0.5)
recv[["rcvTime", "sender", "x", "y", "speed"]].head()
```

```python
print("Different cars heard from:", recv["sender"].nunique())
recv.groupby("sender").size().sort_values(ascending=False)
```
